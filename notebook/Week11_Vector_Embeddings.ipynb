```python
import os
import pandas as pd
import numpy as np
import chromadb
from sentence_transformers import SentenceTransformer

print("Week 11 - Vector Embeddings with ChromaDB")
print("Libraries imported successfully!")
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

print("Sentence Transformer model loaded successfully!")
```

```python
client = chromadb.PersistentClient(path="../chroma_db")

collection = client.get_or_create_collection(
    name="ai_documents"
)

print("ChromaDB connected successfully!")
print("Total documents:", collection.count())
```

```python
query = "How do computers learn from data?"

query_embedding = model.encode([query]).tolist()

results = collection.query(
    query_embeddings=query_embedding,
    n_results=5
)

print("Top 5 Similar Documents:")

for i, doc in enumerate(results["documents"][0]):
    print(i + 1, doc)
```

```python
query = "Artificial Intelligence"

print("Keyword-Based Search Results:")

for filename in os.listdir("../dataset/documents"):
    if filename.endswith(".txt"):
        with open("../dataset/documents/" + filename, "r", encoding="utf-8") as file:
            content = file.read()

        if query.lower() in content.lower():
            print(filename, ":", content)
```

```python
print("SEARCH COMPARISON")

print("1. Keyword Search finds documents containing exact words.")
print("2. Semantic Search finds documents based on meaning.")
print("3. ChromaDB retrieves the Top-5 relevant documents.")
print("4. Semantic Search can find relevant content even without exact keyword matches.")
```
